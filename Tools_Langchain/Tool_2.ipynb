```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class Multiply(BaseModel):
    a: float = Field(..., description="The first number to multiply.")
    b: float = Field(..., description="The second number to multiply.")
```

```python
def multiply_numbers(a: float, b: float) -> float:
    """Multiplies two numbers together."""
    return a * b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_numbers,
    name="Multiply",
    description="Multiplies two numbers together.",
    args_schema=Multiply
)
```

```python
result = multiply_tool.invoke({"a": 5, "b": 3})
result
```

```python
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
print(multiply_tool.args_schema.model_json_schema())
```
